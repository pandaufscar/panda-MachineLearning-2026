Ex 1 - criando e lendo arquivo frutas.txt

```python
# 1. Criando o arquivo frutas.txt e escrevendo os nomes
frutas = ["Maçã", "Banana", "Uva", "Morango", "Abacaxi"]

with open("frutas.txt", "w", encoding="utf-8") as arquivo:
    for fruta in frutas:
        arquivo.write(fruta + "\n")

print("Arquivo 'frutas.txt' criado com sucesso!\n")
```

```python
# 2. Lendo o arquivo e exibindo em maiúsculas
with open("frutas.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        print(linha.strip().upper())
```

Ex 2

```python
# 1. Gerando a lista usando List Comprehension
linhas = [f"O cubo de {x} é {x**3}\n" for x in range(1, 51)]

# 2. Gravando a lista de uma vez só no arquivo
with open("cubos.txt", "w", encoding="utf-8") as arquivo:
    arquivo.writelines(linhas)
print("Arquivo 'cubos.txt' gerado com sucesso!")

with open("cubos.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        print(linha.strip())
```

Ex 3

```python
# 1. Criando o arquivo inicial idades.txt
idades = ["15", "22", "12", "18", "30", "45", "17", "19", "50", "10"]

with open("idades.txt", "w", encoding="utf-8") as f:
    # Escreve cada idade em uma linha
    f.write("\n".join(idades))

# 2. Lendo, filtrando com List Comprehension e salvando em um novo arquivo
with open("idades.txt", "r", encoding="utf-8") as f:
    maiores = [linha for linha in f if int(linha.strip()) > 18]

# 3. Salvando o resultado no arquivo maiores.txt
with open("maiores.txt", "w", encoding="utf-8") as f:
    f.writelines(maiores)

# 4. Checando se foi salvo corretamente
with open("maiores.txt", "r") as f:
    print(f.read())
```
