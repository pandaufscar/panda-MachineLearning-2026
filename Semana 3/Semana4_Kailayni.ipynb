# Exercício 1 - Leitura de arquivo

```python
# Exercício 1: Ler arquivo de frutas e exibir em maiúsculas

# Primeiro, vamos criar o arquivo frutas.txt
frutas = ["maçã\n", "banana\n", "laranja\n", "uva\n", "manga\n"]

with open('frutas.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.writelines(frutas)

# Agora vamos ler o arquivo e exibir cada fruta em maiúsculas
print("=== Frutas em maiúsculas ===")
with open('frutas.txt', 'r', encoding='utf-8') as arquivo:
    for linha in arquivo:
        fruta = linha.strip()  # Remove o \n
        print(fruta.upper())
```

# Exercício 2 - Escrita de arquivo

```python
# Exercício 2: Gerar números de 1 a 50, calcular o cubo e salvar em arquivo

with open('cubos.txt', 'w', encoding='utf-8') as arquivo:
    for x in range(1, 51):
        cubo = x ** 3
        arquivo.write(f"O cubo de {x} é {cubo}\n")

# Vamos verificar o resultado lendo as primeiras 10 linhas
print("\n=== Primeiras 10 linhas do arquivo cubos.txt ===")
with open('cubos.txt', 'r', encoding='utf-8') as arquivo:
    for i, linha in enumerate(arquivo):
        if i < 10:  # Mostra apenas as 10 primeiras linhas
            print(linha.strip())
```

# Exercício 3 - Integrado (arquivo + list comprehension)

```python
# Exercício 3: Filtrar idades maiores que 18 usando list comprehension

# Primeiro, vamos criar o arquivo idades.txt com 10 idades
idades = [15, 22, 17, 19, 25, 16, 30, 18, 21, 14]

with open('idades.txt', 'w', encoding='utf-8') as arquivo:
    for idade in idades:
        arquivo.write(f"{idade}\n")

# Ler o arquivo, filtrar idades > 18 e salvar em maiores.txt
with open('idades.txt', 'r', encoding='utf-8') as arquivo:
    # Ler todas as idades e converter para inteiro
    todas_idades = [int(linha.strip()) for linha in arquivo]

# Filtrar apenas idades maiores que 18
idades_maiores = [idade for idade in todas_idades if idade > 18]

# Salvar o resultado em maiores.txt
with open('maiores.txt', 'w', encoding='utf-8') as arquivo:
    for idade in idades_maiores:
        arquivo.write(f"{idade}\n")

# Exibir resultados
print("\n=== Todas as idades ===")
print(todas_idades)

print("\n=== Idades maiores que 18 ===")
print(idades_maiores)

print("\n=== Conteúdo do arquivo maiores.txt ===")
with open('maiores.txt', 'r', encoding='utf-8') as arquivo:
    for linha in arquivo:
        print(linha.strip())
```
