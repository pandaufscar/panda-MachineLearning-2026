#Exercício 1 – Leitura de arquivo
Crie um arquivo chamado frutas.txt com 5 nomes de frutas (um por linha). Escreva um programa que leia o arquivo e exiba cada fruta em maiúsculas.


```python
# Criando o arquivo frutas.txt
with open("frutas.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("maçã\n")
    arquivo.write("banana\n")
    arquivo.write("uva\n")
    arquivo.write("laranja\n")
    arquivo.write("morango\n")

# Lendo o arquivo e exibindo em maiúsculas
with open("frutas.txt", "r", encoding="utf-8") as arquivo:
    for fruta in arquivo:
        print(fruta.strip().upper())
```

#Exercício 2 – Escrita de arquivo
Crie um programa que gere os números de 1 a 50, calcule o cubo de cada um, e salve no
formato "O cubo de X é Y" em um arquivo chamado cubos.txt.


```python
# Criando o arquivo cubos.txt
with open("cubos.txt", "w", encoding="utf-8") as arquivo:

    # Gerando números de 1 a 50
    for numero in range(1, 51):

        # Calculando o cubo
        cubo = numero ** 3

        # Salvando no arquivo
        arquivo.write(f"O cubo de {numero} é {cubo}\n")

print("Arquivo cubos.txt criado com sucesso!")
```

#Exercício 3 – Escrita de arquivo
Crie um arquivo idades.txt com 10 idades (uma por linha). Leia o arquivo, use list
comprehension para filtrar apenas as idades maiores que 18, e salve o resultado em um
arquivo maiores.txt.


```python
# Criando o arquivo idades.txt
with open("idades.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("12\n")
    arquivo.write("25\n")
    arquivo.write("18\n")
    arquivo.write("30\n")
    arquivo.write("15\n")
    arquivo.write("42\n")
    arquivo.write("19\n")
    arquivo.write("17\n")
    arquivo.write("60\n")
    arquivo.write("21\n")

# Lendo as idades do arquivo
with open("idades.txt", "r", encoding="utf-8") as arquivo:

    # List comprehension para converter em inteiro
    idades = [int(linha.strip()) for linha in arquivo]

# Filtrando apenas maiores de 18
maiores = [idade for idade in idades if idade > 18]

# Criando o arquivo maiores.txt
with open("maiores.txt", "w", encoding="utf-8") as arquivo:

    for idade in maiores:
        arquivo.write(f"{idade}\n")

print("Arquivo maiores.txt criado com sucesso!")
```
