Exercício 1 – Leitura de arquivo
Crie um arquivo chamado frutas.txt com 5 nomes de frutas (um por linha). Escreva um
programa que leia o arquivo e exiba cada fruta em maiúsculas.

```python
with open('frutas.txt', 'r') as arquivo:
    for linha in arquivo:
        fruta = linha.strip()
        print(fruta.upper())

```

Exercício 2 – Escrita de arquivo
Crie um programa que gere os números de 1 a 50, calcule o cubo de cada um, e salve no
formato "O cubo de X é Y" em um arquivo chamado cubos.txt.

```python
numeros = list(range(1, 51))

cubo = []

for numero in numeros:
    cubo.append(numero ** 3)

with open('cubos.txt', 'w') as arquivo:
    for i in range(len(numeros)):
        arquivo.write(f'O cubo de {numeros[i]} é {cubo[i]}\n')
```

Exercício 3 – Integrado (arquivo + list comprehension)
Crie um arquivo idades.txt com 10 idades (uma por linha). Leia o arquivo, use list
comprehension para filtrar apenas as idades maiores que 18, e salve o resultado em um
arquivo maiores.txt.

```python
with open('idades.txt', 'r') as arquivo:
  idades = [int(linha.strip()) for linha in arquivo]

maiores = [idade for idade in idades if idade >= 18]

with open('maiores.txt', 'w') as saida:
  for idade in maiores:
    saida.write(f'{idade}\n')
```
