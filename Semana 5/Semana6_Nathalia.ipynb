Identificar e tratar valores nulos

```python
import pandas as pd

titanic = pd.read_csv("Titanic-Dataset.csv")

titanic.head(5)
```

```python
# Agrupamentos com groupby
titanic.groupby('Sex')

# Quantas pessoas estão em cada gênero?
titanic.groupby('Sex').size()
```

```python
# Operações com merge
pessoa = titanic[['PassengerId', 'Name', 'Age']]
embarque = titanic[['PassengerId', 'Embarked', 'Fare', 'Survived']]
merge = pd.merge(pessoa,embarque, on='PassengerId')

print(merge.head())
```

```python
# Tratar valores nulos
titanic.isnull().sum()
# Há alguns valores nulos nas colunas
# Age, Cabin e Embarked.
```

```python
# Assim, para tratar esse dataset vou fazer:
# Preencher Age com a média, pois é uma variável numérica.
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].mean())

# Preencher Cabin com um valor padrão, indicando que
# a cabine do passageiro é desconhecida.
titanic['Cabin'] = titanic['Cabin'].fillna('Unknown')

# Preencher Embarked com a moda, pois há poucos
# valores ausentes e é uma variável categórica.
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
```

```python
#Verificando se restam valores nulos:
titanic.isnull().sum()
# Como não há, a limpeza foi concluida.
```
