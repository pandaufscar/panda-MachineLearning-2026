#Importação

```python
import pandas as pd
```

# Criação dos arquivos CSV

```python
with open("alunos.csv", "w", encoding="utf-8") as arquivo:
    arquivo.write("id,nome,idade,nota,id_curso\n")
    arquivo.write("1,Ana,20,8.5,101\n")
    arquivo.write("2,Bruno,22,6.0,102\n")
    arquivo.write("3,Carla,19,,101\n")
    arquivo.write("4,Diego,21,7.5,103\n")
    arquivo.write("5,Elaine,23,5.0,\n")
    arquivo.write("6,Felipe,20,9.0,102\n")

with open("cursos.csv", "w", encoding="utf-8") as arquivo:
    arquivo.write("id_curso,curso\n")
    arquivo.write("101,Computação\n")
    arquivo.write("102,Engenharia\n")
    arquivo.write("103,Matemática\n")
```

# Leitura dos arquivos

```python
alunos = pd.read_csv("alunos.csv")
cursos = pd.read_csv("cursos.csv")
```

# Visualização

```python
print(alunos)

print("\n")

print(cursos)
```

# Groupby

```python
media_curso = alunos.groupby("id_curso")["nota"].mean()

print(media_curso)
```

# Merge

```python
dados_completos = pd.merge(alunos, cursos, on="id_curso")

print(dados_completos)
```

#Valores nulos

```python
print(alunos.isnull().sum())
```

#Tratamento de nulos

```python
media_notas = alunos["nota"].mean()

alunos["nota"] = alunos["nota"].fillna(media_notas)

alunos["id_curso"] = alunos["id_curso"].fillna(0)

print(alunos)
```
