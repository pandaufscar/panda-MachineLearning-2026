# 1 - Instalar bibliotecas e  criação do dataset

```python
import pandas as pd
import numpy as np

dados_produtos = {
    'produto_id': [101, 102, 103, 104, 105],
    'nome_produto': ['Notebook', 'Smartphone', 'Monitor 4K', 'Teclado Mecânico', 'Mouse Sem Fio'],
    'categoria': ['Eletrônicos', 'Eletrônicos', 'Monitores', 'Acessórios', 'Acessórios'],
    'preco_unitario': [4500.00, 2500.00, 1800.00, 350.00, 150.00]
}
df_produtos = pd.DataFrame(dados_produtos)

dados_vendas = {
    'venda_id': [1, 2, 3, 4, 5, 6, 7],
    'produto_id': [101, 102, 101, 104, 103, 105, 102],
    'quantidade': [2, 1, np.nan, 3, 1, 5, np.nan],
    'loja': ['SP', 'RJ', 'SP', 'MG', 'RJ', 'SP', 'MG']
}
df_vendas = pd.DataFrame(dados_vendas)

df_produtos.to_csv('produtos.csv', index=False)
df_vendas.to_csv('vendas.csv', index=False)
print("Tabelas 'produtos.csv' e 'vendas.csv' criadas com sucesso!")
```

# 2 - Identificação e tratamento de valores nulos

```python
vendas = pd.read_csv('vendas.csv')
produtos = pd.read_csv('produtos.csv')

print("Valores nulos em Vendas:\n", vendas.isnull().sum())

mediana_qtd = vendas['quantidade'].median()
vendas['quantidade'] = vendas['quantidade'].fillna(mediana_qtd)

print("\nApós o tratamento (valores nulos remanescentes):")
print(vendas.isnull().sum())
```

# 3 - Merge

```python
df_consolidado = pd.merge(vendas, produtos, on='produto_id', how='inner')

df_consolidado['faturamento_total'] = df_consolidado['quantidade'] * df_consolidado['preco_unitario']

df_consolidado.head()
```

# 4 - Groupby

```python
agrupado_categoria = df_consolidado.groupby('categoria').agg({
    'quantidade': 'sum',
    'faturamento_total': 'sum'
}).reset_index()

print("Resumo por Categoria")
print(agrupado_categoria)

agrupado_loja = df_consolidado.groupby('loja')['faturamento_total'].sum().reset_index()

print("\n--- Resumo por Loja ---")
print(agrupado_loja)
```
