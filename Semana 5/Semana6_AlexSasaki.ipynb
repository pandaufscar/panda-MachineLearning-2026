```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sriharshabsprasad/movielens-dataset-100k-ratings")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df_movies = pd.read_csv(path + "/ml-latest-small/movies.csv")
df_ratings = pd.read_csv(path + "/ml-latest-small/ratings.csv")
```

```python
display(df_movies.head())
display(df_ratings.head())
```

```python
# Primeiro faremos o Merge dos dois datasets que interessam:
# Uniremos o dataset com os dados dos filmes, e o dataset com as notas:

df_movies_rating = pd.merge(df_movies, df_ratings, on="movieId")
df_movies_rating.head()
```

```python
# Agora, agruparemos as notas por filme filme e faremos uma média:

df_movies_rating.groupby(["movieId", "title"])["rating"].mean()
```

```python
# Verificaremos se existem valores nulos

df_movies_rating.isnull().sum()

# Como não existem valores nulos, não há necessidade de tratamento.
```
