Foi utilizado banco de dados Olist de vendas:

```python
import pandas as pd

# Carregando tabelas
df_orders = pd.read_csv('olist_orders_dataset.csv')
df_customers = pd.read_csv('olist_customers_dataset.csv')

# Juntando com merge
df_merged = pd.merge(df_orders, df_customers, on='customer_id', how='inner')
display(df_merged.head())
```

```python
# Identificando e tratando valores nulos
print("Quantidade de valores nulos por coluna:")
print(df_merged.isnull().sum())
df_cleaned = df_merged.dropna(subset=['order_approved_at']) # Limpando nulos da coluna order_approved_at
print(df_cleaned.isnull().sum())
```

```python
# Usando groupby
pedidos_por_estado = df_cleaned.groupby('customer_state')['order_id'].count().reset_index()

pedidos_por_estado = pedidos_por_estado.rename(columns={'order_id': 'total_pedidos'})

pedidos_por_estado = pedidos_por_estado.sort_values(by='total_pedidos', ascending=False)

print("\nTop 5 Estados com mais pedidos aprovados:")
print(pedidos_por_estado.head())
```
