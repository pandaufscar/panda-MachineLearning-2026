```python
# Exercício 3
def maior_da_lista(lista):
    if not lista:
        return None

    maior = lista[0]

    for n in lista:
        if n > maior:
            maior = n

    return maior

lista = [5, 3, 2, 7, 6, 5, 8, 0]
print(maior_da_lista(lista))
```
