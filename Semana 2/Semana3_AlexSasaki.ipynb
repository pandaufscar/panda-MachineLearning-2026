Exercício 1: Faça um programa
que pede um número e diz se é
par ou ímpar (if/else)

```python
numero = input("Insira um número: ")

if int(numero) % 2 == 0:
  print("É par")
else:
  print("É impar")
```

Exercício 2: Use um loop for para
somar todos os números de 1 a
100.

```python
soma = 0
for i in range (1, 101):
  soma = i + soma
print(soma)
```

Exercício 3: Crie uma função que
recebe uma lista e retorna o
maior valor

```python
import numpy as np
def maior_valor(lista):
  maior = lista[0]
  for i in lista:
    if i > maior:
      maior = i
  return maior

lista = np.random.randint(1, 100, 10)
print(lista)
print(maior_valor(lista))
```
