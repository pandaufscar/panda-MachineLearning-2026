```python
soma = 0

for numero in range(1, 101):
    soma += numero

print(f"A soma de todos os números de 1 a 100 é: {soma}")
```

```python
def encontrar_maior(lista):
    maior = lista[0]
    for numero in lista:
        if numero > maior:
            maior = numero

    return maior

# Testando a função
numeros = [15, 42, 7, 88, 23, 91, 4]
resultado = encontrar_maior(numeros)

print(f"O maior valor da lista é: {resultado}")
```
