# Exercício 1 - Par ou Ímpar



```python
numero = int(input("Digite um número: "))

if numero % 2 == 0:
    print(f"O número {numero} é PAR")
else:
    print(f"O número {numero} é ÍMPAR")
```

# Exercício 2 - Soma de 1 a 100

```python
soma = 0

for i in range(1, 101):
    soma += i

print(f"A soma dos números de 1 a 100 é: {soma}")
```

# Exercício 3 - Maior valor de uma lista

```python
def encontrar_maior(lista):
    if not lista:  # verifica se a lista está vazia
        return None
    maior = lista[0]
    for valor in lista:
        if valor > maior:
            maior = valor
    return maior

# Exemplo de uso
minha_lista = [3, 8, 2, 10, 5, 7]
resultado = encontrar_maior(minha_lista)
print(f"O maior valor da lista {minha_lista} é: {resultado}")
```
