Exercício 1: Faça um programa
que pede um número e diz se é
par ou ímpar (if/else)

```python
numero = int(input("Digite um número "))

if numero % 2 == 0:
  print("Esse número é par")
else:
  print("Esse número é ímpar")
```

Exercício 2: Use um loop for para
somar todos os números de 1 a
100.

```python
soma = 0

for i in range (1,101):
  soma = i + soma



print(soma)
```

Exercício 3: Crie uma função que
recebe uma lista e retorna o
maior valor

```python
def maior_valor(lista):
   return max(lista)

lista_numeros = [1,2,3,4,5]
print("O maior valor é: ",maior_valor(lista_numeros))
```
