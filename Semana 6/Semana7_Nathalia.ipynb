```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

titanic = pd.read_csv("Titanic-Dataset.csv")
```

```python
titanic.head(5)
```

1. Histograma


```python
titanic['Age'].hist(bins=30)
plt.title('Frequência das Idades')
plt.xlabel('Idade do indivíduo')
plt.ylabel('Quantidade de Passageiros')
plt.show()
```

No histograma é possível ver que a maioria das idades está entre 20 anos a 37 anos. As idades menos frequentes são próximas dos 70 anos e dos 80.

2. Scatter Plot:

```python
plt.scatter(titanic['Age'], titanic['Fare'])
plt.title('Idade x Valor')
plt.xlabel('Idade')
plt.ylabel('Valor que pagou')
plt.show()
```

A idade das pessoas e o valor que elas pagaram é bem disperso, com a faixa etária de 20 a 30 anos sendo bem concentrada no valor mais baixo. Há 2 pessoas bem próximas dos 40 anos que pagaram um valor bem alto, os únicos que pagaram esse valor no dataset.

3. Boxplot

```python
titanic.boxplot(column='Age')
plt.title('Boxplot das Idades')
plt.ylabel('Idade')
plt.show()
```

Q1 encontra-se nos 20 anos, isso significa que 25% dos indivíduos possuem essa idade ou menos. A mediana é próxima dos 30 anos, que também é o Q2 e significa que 50% se encontram nessa idade ou menos. Por fim, o Q3 está próximo dos 40 anos e isso significa que 75% está com até essa idade. Os outliers estão presentes depois dos 60 anos, se concentrando nos 70 anos.
