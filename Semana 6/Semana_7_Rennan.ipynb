```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Carregando tabela
df_items = pd.read_csv('olist_order_items_dataset.csv')
df_filtrado = df_items[df_items['price'] < 500]
sns.set_theme(style="whitegrid")
```

```python
# Histograma
plt.figure(figsize=(10, 5))
sns.histplot(df_filtrado['price'], bins=40, kde=True, color='purple')
plt.title('Distribuição de Preços dos Produtos (até R$ 500)')
plt.xlabel('Preço (R$)')
plt.ylabel('Quantidade de Vendas')
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_filtrado.sample(1000), x='price', y='freight_value', alpha=0.6)
plt.title('Relação entre Preço do Produto e Valor do Frete')
plt.xlabel('Preço (R$)')
plt.ylabel('Valor do Frete (R$)')
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(x=df_filtrado['freight_value'], color='cyan')
plt.title('Boxplot do Valor de Frete')
plt.xlabel('Valor do Frete (R$)')
plt.show()
```

1. Histograma

O gráfico mostra uma forte concentração de vendas em produtos baratos.

A maioria absoluta dos itens comercializados custa menos de R$ 100,00.

2. Scatter Plot

A falta de um padrão visual indica que não existe relação direta entre o preço do produto e o valor do frete.

Isso sugere que o custo da entrega depende mais de outros fatores, como o peso do pacote.

3. Boxplot

A linha central da caixa mostra que a maior parte dos fretes custa em torno de R\$ 15,00 a R\$ 20,00.

Os pontos isolados à direita representam as entregas que tiveram valores atipicamente caros.
