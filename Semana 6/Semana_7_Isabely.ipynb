# Importação das bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Carregando o dataset

```python
tips = sns.load_dataset("tips")

print(tips.head())
```

# Histograma
**Objetivo:**
*Visualizar a distribuição das gorjetas.*

```python
plt.figure(figsize=(8,5))

plt.hist(tips["tip"], bins=10)

plt.title("Distribuição das Gorjetas")
plt.xlabel("Valor da Gorjeta")
plt.ylabel("Frequência")

plt.show()
```

## Interpretação do Histograma

- A maior parte das gorjetas está concentrada entre 2 e 4 dólares.  
- Valores muito altos de gorjeta aparecem com menor frequência.  
- A distribuição apresenta assimetria à direita, indicando poucos valores extremos.

# Scatter Plot
**Objetivo:**
*Analisar relação entre valor da conta e gorjeta.*

```python
plt.figure(figsize=(8,5))

plt.scatter(tips["total_bill"], tips["tip"])

plt.title("Conta Total x Gorjeta")
plt.xlabel("Valor da Conta")
plt.ylabel("Gorjeta")

plt.show()
```

## Interpretação do Scatter Plot

- Existe uma tendência de aumento da gorjeta conforme o valor da conta aumenta.  
- Os pontos apresentam dispersão moderada, mas indicam correlação positiva entre as variáveis.  
- Contas maiores geralmente estão associadas a gorjetas maiores.

#Boxplot
**Objetivo:**
*Comparar distribuição de gorjetas entre fumantes e não fumantes.*

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="smoker", y="tip", data=tips)

plt.title("Gorjetas entre Fumantes e Não Fumantes")

plt.show()
```

## Interpretação do Boxplot

- Os grupos apresentam medianas relativamente próximas.  
- Existem valores extremos em ambos os grupos, representando gorjetas muito altas.  
- A variabilidade das gorjetas entre fumantes parece ligeiramente maior.
