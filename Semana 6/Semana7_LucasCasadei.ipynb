# 1 - Importar bibliotecas e visualização

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("penguins")
df = df.dropna()
df.head()

print(df.info())
print(df.describe())
print(df["species"].value_counts())
print(df.isnull().sum())
```

```python
df.head()
```

# 2 - Histograma

```python
plt.figure(figsize=(9,5))

sns.histplot(
    df["flipper_length_mm"],
    bins=20,
    kde=True
)

plt.axvline(
    df["flipper_length_mm"].mean(),
    linestyle="--",
    label="Média"
)

plt.axvline(
    df["flipper_length_mm"].median(),
    linestyle=":",
    label="Mediana"
)

plt.title("Distribuição do comprimento das nadadeiras")
plt.xlabel("Comprimento da nadadeira (mm)")
plt.ylabel("Frequência")

plt.legend()

plt.show()
```

O gráfico apresenta uma distribuição bimodal, sendo o primeiro pico concentrando-se em torno de 190 mm e o segundo próximo a 215 mm. Isso sugere que o conjunto de dados é formado por grupos distintos de pinguins com características físicas diferentes (espécies menores vs. espécies maiores).

# 3 - Scatter Plot

```python
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    size="body_mass_g",
    alpha=0.7
)

plt.title("Comprimento x profundidade do bico")
plt.xlabel("Comprimento do bico (mm)")
plt.ylabel("Profundidade do bico (mm)")

plt.grid(True)

plt.show()
```

```python
sns.lmplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    height=6
)

plt.title("Relação entre comprimento e profundidade do bico")

plt.show()
```

Embora a correlação global pareça negativa e confusa à primeira vista, a separação por cores (hue='species') revela três agrupamentos perfeitamente definidos e uma correlação positiva interna.
Dentro de cada espécie isolada, o aumento no comprimento do bico acompanha o aumento da sua profundidade, tornando estas dimensões uma excelente assinatura biológica para identificar as espécies.

# 4 - Boxplot

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    data=df,
    x="species",
    y="body_mass_g"
)

sns.stripplot(
    data=df,
    x="species",
    y="body_mass_g",
    color="black",
    alpha=0.4
)

plt.title("Distribuição da massa corporal por espécie")
plt.xlabel("Espécie")
plt.ylabel("Massa corporal (g)")

plt.grid(True)

plt.show()
```

O gráfico revela uma  disparidade física entre as espécies, destacando visualmente os pinguins Gentoo com valores de mediana e caixas em patamares muito mais elevados, ou seja, com maior massa corporal e nadadeiras. Em contrapartida, as espécies Adelie e Chinstrap apresentam distribuições e medianas bastante semelhantes entre si, ambas exibindo uma variabilidade controlada e baixa presença de outliers após o tratamento dos dados.
