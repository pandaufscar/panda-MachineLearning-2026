# Célula 1 — Importar bibliotecas

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# carregar dataset
tips = sns.load_dataset("tips")

# visualizar primeiras linhas
tips.head()
```

# Célula 2 — Histograma

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=tips,
    x='total_bill',
    bins=20,
    kde=True
)

plt.title("Histograma: Distribuição do Valor Total da Conta")
plt.xlabel("Valor da Conta (USD)")
plt.ylabel("Frequência")

plt.show()


print("="*50)
print("INTERPRETAÇÃO")
print("="*50)

print("""
O histograma mostra a distribuição dos valores das contas no restaurante.
A maior concentração está entre aproximadamente 10 e 20 dólares.
Também existem poucos valores muito altos, indicando uma distribuição
assimétrica à direita.
""")
```

# Célula 3 — Scatter Plot

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time'
)

plt.title("Scatter Plot: Conta Total x Gorjeta")
plt.xlabel("Conta Total (USD)")
plt.ylabel("Gorjeta (USD)")

plt.show()


print("="*50)
print("INTERPRETAÇÃO")
print("="*50)

print("""
O gráfico mostra uma relação positiva entre conta total e gorjeta.
Em geral, contas maiores tendem a gerar gorjetas maiores.
Apesar disso, existe dispersão nos pontos, indicando comportamentos
diferentes entre os clientes.
""")
```

# Célula 4 — Boxplot

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=tips,
    x='day',
    y='tip'
)

media = tips['tip'].mean()

plt.axhline(
    media,
    color='red',
    linestyle='--',
    label=f"Média Geral: ${media:.2f}"
)

plt.legend()

plt.title("Boxplot: Distribuição de Gorjetas por Dia")
plt.xlabel("Dia")
plt.ylabel("Valor da Gorjeta (USD)")

plt.show()


print("="*50)
print("INTERPRETAÇÃO")
print("="*50)

print("""
O boxplot mostra que domingo apresenta a maior mediana de gorjeta,
indicando valores centrais mais altos em comparação aos outros dias.

O domingo também possui maior amplitude interquartil, sugerindo maior
variabilidade nas gorjetas recebidas.

O sábado apresenta diversos outliers, indicando a ocorrência de
gorjetas excepcionalmente altas em alguns casos.
""")
```
