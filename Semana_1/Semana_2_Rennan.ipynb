```python
amigos = [
    {"nome": "Ana", "idade": 25, "cidade": "São Paulo"},
    {"nome": "Bruno", "idade": 30, "cidade": "Curitiba"},
    {"nome": "Carla", "idade": 22, "cidade": "Salvador"}
]

print("--- Lista de Amigos ---")
for amigo in amigos:
    print(f"Nome: {amigo['nome']} | Idade: {amigo['idade']} | Cidade: {amigo['cidade']}")
```
