# Exercício 1

Crie uma lista com 5 itens de supermercado.
Depois:

*   Adicione mais um item
*   Mostre o terceiro item da lista




```python
supermercado = ['arroz', 'feijao', 'batatas', 'macarrao', 'leite'] # Criação da lista
supermercado.append('cafe') # Adição de mais um item ao final da lista
print(supermercado[2]) # Print do terceiro item, sendo o índice 2 o terceiro.
```
