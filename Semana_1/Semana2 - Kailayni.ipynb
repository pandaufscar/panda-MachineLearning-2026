# Exercício 1 - Lista de compras

```python
# Criando lista com 5 itens de supermercado
lista_compras = ['Arroz', 'Feijão', 'Macarrão', 'Leite', 'Café']

# Exibindo a lista original
print("Lista de compras original:")
print(lista_compras)

# Adicionando mais um item
lista_compras.append('Açúcar')
print("\nLista após adicionar mais um item:")
print(lista_compras)

# Mostrando o terceiro item da lista (índice 2)
print(f"\nO terceiro item da lista é: {lista_compras[2]}")
```

# Exercício 2 - Cadastro de aluno

```python
# Criando dicionário do aluno
aluno = {
    'nome': 'João Silva',
    'idade': 20,
    'curso': 'Engenharia de Software',
    'notas': [8.5, 7.0, 9.5]
}

# Calculando a média das notas
media = sum(aluno['notas']) / len(aluno['notas'])

# Mostrando o nome e a média
print(f"Nome do aluno: {aluno['nome']}")
print(f"Média das notas: {media:.2f}")

# Exibindo todas as informações
print("\n--- Dados completos do aluno ---")
print(f"Nome: {aluno['nome']}")
print(f"Idade: {aluno['idade']}")
print(f"Curso: {aluno['curso']}")
print(f"Notas: {aluno['notas']}")
print(f"Média: {media:.2f}")
```

# Exercício 3 - Informações de 3 amigos

```python
# Criando uma lista de dicionários com informações de 3 amigos
amigos =[
    {
        'nome': 'Ana Oliveira',
        'idade': 25,
        'cidade': 'São Paulo'
    },
    {
        'nome': 'Carlos Santos',
        'idade': 28,
        'cidade': 'Rio de Janeiro'
    },
    {
        'nome': 'Mariana Costa',
        'idade': 23,
        'cidade': 'Belo Horizonte'
    }
]

# Exibindo todas as informações dos amigos
print("Informações dos amigos:")
print("-" * 40)

for i, amigo in enumerate(amigos, 1):
  print(f"\nAmigo {i}:")
  print(f"Nome: {amigo['nome']}")
  print(f"Idade: {amigo['idade']} anos")
  print(f"Cidade: {amigo['cidade']}")

# Exibindo apenas os nomes
print("\n" + "=" * 40)
print("Apenas os nomes dos amigos:")
for amigo in amigos:
  print(f"- {amigo['nome']}")
```
