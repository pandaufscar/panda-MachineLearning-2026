# Exercício 1 - Lista de Compras
Crie uma lista com 5 itens de supermercado.
Depois:

*   Adicione mais um item
*   Mostre o terceiro item da lista




```python
supermercado = ['arroz', 'feijao', 'leite', 'macarrao', 'biscoito'] # Criação da lista
supermercado.append('cafe') # Adição de mais um item ao final da lista
print(supermercado[2]) # Print do terceiro item, que está no índice 2
```
