# Exercício Prático - Análise de dados de alunos

```python
import pandas as pd
import numpy as np

# ==================== PASSO 1: Criar o dataset alunos.csv ====================
print("=" * 60)
print("CRIANDO O DATASET alunos.csv")
print("=" * 60)

# Dados fictícios dos alunos
dados = {
    'nome': ['Ana Silva', 'João Santos', 'Maria Oliveira', 'Pedro Costa',
             'Julia Lima', 'Carlos Souza', 'Fernanda Rocha', 'Lucas Pereira',
             'Juliana Dias', 'Rafael Alves', 'Camila Nunes', 'Bruno Mendes',
             'Patrícia Araujo', 'André Cardoso', 'Beatriz Martins'],
    'idade': [20, 22, 19, 21, 23, 20, 22, 21, 24, 20, 19, 22, 23, 21, 20],
    'nota': [8.5, 6.0, 9.0, 7.5, 5.5, 8.0, 9.5, 6.5, 7.0, 8.5, 9.0, 6.0, 7.5, 8.0, 9.5],
    'curso': ['Engenharia', 'Medicina', 'Computação', 'Engenharia', 'Medicina',
              'Computação', 'Engenharia', 'Medicina', 'Computação', 'Engenharia',
              'Medicina', 'Computação', 'Engenharia', 'Medicina', 'Computação']
}

# Criando DataFrame
df = pd.DataFrame(dados)

# Salvando em arquivo CSV
df.to_csv('alunos.csv', index=False, encoding='utf-8')
print("Arquivo 'alunos.csv' criado com sucesso!")
print(f"Total de alunos: {len(df)}")
print("\nPrimeira visualização dos dados criados:")
print(df.head(3))

# ==================== TAREFA 1: Carregar o arquivo CSV ====================
print("\n" + "=" * 60)
print("TAREFA 1 - Carregando o arquivo CSV com Pandas")
print("=" * 60)

# Carregando o arquivo CSV
alunos_df = pd.read_csv('alunos.csv', encoding='utf-8')
print("✓ Arquivo 'alunos.csv' carregado com sucesso!")
print(f"✓ Shape do DataFrame: {alunos_df.shape}")  # (linhas, colunas)
print(f"✓ Colunas disponíveis: {list(alunos_df.columns)}")

# ==================== TAREFA 2: Mostrar as primeiras 5 linhas ====================
print("\n" + "=" * 60)
print("TAREFA 2 - Primeiras 5 linhas do dataset")
print("=" * 60)

print(alunos_df.head(5))

# ==================== TAREFA 3: Usar info() e describe() ====================
print("\n" + "=" * 60)
print("TAREFA 3 - Informações gerais (info() e describe())")
print("=" * 60)

print("\n--- Informações do DataFrame (info()) ---")
print(alunos_df.info())

print("\n--- Estatísticas descritivas (describe()) ---")
print(alunos_df.describe())

# Complemento: mostrar estatísticas para colunas categóricas também
print("\n--- Estatísticas para colunas categóricas (include='object') ---")
print(alunos_df.describe(include=['object']))

# ==================== TAREFA 4: Selecionar apenas as colunas nome e nota ====================
print("\n" + "=" * 60)
print("TAREFA 4 - Selecionando colunas 'nome' e 'nota'")
print("=" * 60)

# Selecionando apenas as colunas nome e nota
nome_nota_df = alunos_df[['nome', 'nota']]
print(nome_nota_df)

print(f"\n✓ Nova seleção tem {len(nome_nota_df)} linhas e {len(nome_nota_df.columns)} colunas")
print(f"✓ Colunas selecionadas: {list(nome_nota_df.columns)}")

# ==================== TAREFA 5: Filtrar alunos com nota >= 7 ====================
print("\n" + "=" * 60)
print("TAREFA 5 - Filtrando alunos com nota >= 7")
print("=" * 60)

# Filtrando alunos com nota maior ou igual a 7
alunos_aprovados = alunos_df[alunos_df['nota'] >= 7]

print(f"Total de alunos com nota >= 7: {len(alunos_aprovados)}")
print(f"Percentual: {(len(alunos_aprovados)/len(alunos_df)*100):.1f}%")
print("\nLista dos alunos aprovados:")
print(alunos_aprovados[['nome', 'nota', 'curso']].to_string(index=False))

# Extra: análises adicionais
print("\n" + "=" * 60)
print("ANÁLISES ADICIONAIS (Complemento)")
print("=" * 60)

# Estatísticas por curso
print("\n--- Média de notas por curso ---")
media_por_curso = alunos_df.groupby('curso')['nota'].mean().round(2)
for curso, media in media_por_curso.items():
    print(f"  {curso}: {media}")

# Distribuição de faixas de nota
print("\n--- Distribuição das notas ---")
faixas = ['0-4', '4-6', '6-7', '7-9', '9-10']
alunos_df['faixa_nota'] = pd.cut(alunos_df['nota'], bins=[0, 4, 6, 7, 9, 10], labels=faixas)
distribuicao = alunos_df['faixa_nota'].value_counts().sort_index()
for faixa, quantidade in distribuicao.items():
    print(f"  {faixa}: {quantidade} aluno(s)")

# Remover a coluna auxiliar criada
alunos_df.drop('faixa_nota', axis=1, inplace=True)

# Melhor aluno por curso
print("\n--- Melhor aluno por curso ---")
melhores_por_curso = alunos_df.loc[alunos_df.groupby('curso')['nota'].idxmax()]
for _, row in melhores_por_curso.iterrows():
    print(f"  {row['curso']}: {row['nome']} - Nota {row['nota']}")

print("\n" + "=" * 60)
print("✓ Todos os exercícios foram concluídos com sucesso!")
print("=" * 60)
```
