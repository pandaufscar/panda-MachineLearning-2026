# Dataset: estudantes.csv (dados fictícios com nome, idade, nota, curso)

Tarefas:
1. Carregar o arquivo CSV com Pandas
2. Mostrar as primeiras 5 linhas
3. Usar info() e describe()
4. Selecionar apenas as colunas nome e
nota
5. Filtrar alunos com nota >= 7

```python
# Importando a biblioteca pandas
import pandas as pd

# Criando o arquivo 'alunos.csv'
with open("alunos.csv", "w", encoding="utf-8") as arquivo:
    arquivo.write("nome,idade,nota,curso\n")
    arquivo.write("Ana,20,8.5,Computação\n")
    arquivo.write("Bruno,22,6.0,Engenharia\n")
    arquivo.write("Carla,19,9.2,Matemática\n")
    arquivo.write("Diego,21,7.0,Física\n")
    arquivo.write("Elaine,23,5.5,Química\n")
    arquivo.write("Felipe,20,8.0,Computação\n")

# 1. Carregar o arquivo CSV com Pandas
df = pd.read_csv("alunos.csv")

# 2. Mostrar as primeiras 5 linhas
print("Primeiras 5 linhas:")
print(df.head())

# 3. Usar info() e describe()
print("\nInformações do DataFrame:")
print(df.info())

print("\nEstatísticas do DataFrame:")
print(df.describe())

# 4. Selecionar apenas as colunas nome e nota
print("\nColunas nome e nota:")
print(df[["nome", "nota"]])

# 5. Filtrar alunos com nota >= 7
print("\nAlunos com nota maior ou igual a 7:")
print(df[df["nota"] >= 7])
```
