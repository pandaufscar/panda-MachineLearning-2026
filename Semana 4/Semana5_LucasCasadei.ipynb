# Cria o arquivo .csv

```python
import pandas as pd
import random

random.seed(42)

nomes = [
    "Ana", "Bruno", "Carlos", "Daniela", "Eduardo", "Fernanda", "Gabriel",
    "Helena", "Igor", "Juliana", "Kleber", "Larissa", "Marcos", "Natália",
    "Otávio", "Patrícia", "Rafael", "Sabrina", "Thiago", "Vanessa"
]

sobrenomes = [
    "Silva", "Souza", "Oliveira", "Pereira", "Costa", "Rodrigues",
    "Almeida", "Nascimento", "Lima", "Araújo"
]

cursos = [
    "Engenharia", "Direito", "Medicina", "Computação", "Administração",
    "Arquitetura", "Economia", "Psicologia"
]

dados = []

for i in range(100):
    nome_completo = f"{random.choice(nomes)} {random.choice(sobrenomes)}"
    idade = random.randint(17, 30)
    nota = round(random.uniform(0, 10), 1)
    curso = random.choice(cursos)

    dados.append({
        "nome": nome_completo,
        "idade": idade,
        "nota": nota,
        "curso": curso
    })

df = pd.DataFrame(dados)

df.to_csv("alunos.csv", index=False)

df.head()
```

# Carrega o csv

```python
df = pd.read_csv("alunos.csv")
df
```
