Ex 1 - criando o arquivo csv

```python
import pandas as pd

# Criando os dados fictícios
dados = {
    "nome": ["Ana", "Bruno", "Carla", "Daniel", "Eduarda", "Fabio", "Gisele", "Hugo", "Igor", "Julia"],
    "idade": [20, 22, 19, 21, 23, 20, 22, 21, 24, 19],
    "nota": [8.5, 6.0, 9.2, 5.5, 7.8, 10.0, 6.5, 8.0, 4.5, 9.0],
    "curso": ["Engenharia", "Direito", "Medicina", "TI", "Artes", "TI", "Engenharia", "Direito", "Artes", "Medicina"]
}

# Transformando em DataFrame e salvando em CSV
df_inicial = pd.DataFrame(dados)
df_inicial.to_csv("alunos.csv", index=False)

print("Arquivo 'alunos.csv' criado com sucesso!")
```

Manipulando dados

```python
# 1. Carregar o arquivo CSV
df = pd.read_csv("alunos.csv")

# 2. Mostrar as primeiras 5 linhas
print("--- Primeiras 5 linhas ---")
df.head()
```

```python

# 3. Usar info() e describe()
print("\n--- Informações do Dataset (info) ---")
df.info()

print("\n--- Resumo Estatístico (describe) ---")
df.describe()
```

```python
# 4. Selecionar apenas as colunas nome e nota
colunas_selecionadas = df[["nome", "nota"]]
print("\n--- Apenas Nome e Nota ---")
colunas_selecionadas.head()
```
