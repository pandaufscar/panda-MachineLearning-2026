```python
import pandas as pd
import numpy as np

df = pd.read_csv('estudantes.csv')

df.head() # Comando que mostra as 5 primeiras linhas
```

```python
df.info() # Mostrando os atributos do dataset

df.describe() # Mostra como os dados estão distribuídos
```

```python
desempenho = df[['Nome', 'Nota']] # Selecionando as colunas nome e nota do dataset

desempenho.head()
```

```python
bom_desempenho = desempenho[desempenho['Nota'] >= 7] # Filtrando o Dataset com notas >=7

bom_desempenho.head(25)
```
